## 📘 Ripasso: la teoria che ti serve

### Java e Python: cambia la grammatica, non il pensiero

| Aspetto | Java | Python |
| --- | --- | --- |
| Tipo delle variabili | dichiarato: `int vita = 100;` | dedotto dal valore: `vita = 100` |
| Controllo dei tipi | prima di partire, dal compilatore | mentre il programma gira |
| Blocchi di codice | parentesi graffe `{ }` | indentazione |
| Fine riga | `;` obbligatorio | niente |
| Stampa | `System.out.println(x);` | `print(x)` |
| Commento | `// testo` | `# testo` |

Python traduce il codice in *bytecode* e poi lo esegue con l'interprete. Anche Java fa così: la differenza è **quando** si controllano gli errori (Java prima di partire, Python mentre gira).

### Una variabile è una freccia

In Python `vita = 100` non dichiara un tipo: <u>ogni variabile è una freccia che punta a un valore</u>.

```text
vita ───▶ 100
nome ───▶ "Aria"
```

Se scrivi `vita = 70`, la freccia si sposta; il valore `100` non è stato modificato. Se scrivi `altra = vita`, ottieni **due frecce verso lo stesso valore**. `True` e `False` hanno la maiuscola.

I **tipi stanno nei valori**, non nelle variabili (*tipizzazione dinamica*): una variabile può puntare prima a un `int` e poi a una `str`.

| Tipo | Che cos'è | Esempio |
| --- | --- | --- |
| `int` | numero intero | `100` |
| `float` | numero con la virgola (nel codice è un punto) | `7.5` |
| `str` | testo | `"Aria"` |
| `bool` | vero o falso | `True`, `False` |

⚠️ `3` e `3.0` sono lo stesso numero ma tipi diversi. E `"3"` è un testo, non un numero. Si controlla con `type()`.

### Input, output e conversioni

`print()` scrive sullo schermo, `input()` legge **sempre testo**. Per fare i conti bisogna convertire con `int()`, `float()`, `str()`.

```python
forza = int(input("Forza: "))   # senza int(), forza + 1 darebbe TypeError
```

Una **f-string** è un testo che comincia con `f` e contiene valori fra graffe:

```python
nome = "Aria"
vita = 70
print(f"{nome} ha {vita} punti vita.")            # Aria ha 70 punti vita.
print(f"{nome} ha {vita / 100:.0%} della vita.")  # Aria ha 70% della vita.
```

Dopo i due punti si sceglie il formato: `:.0%` stampa una percentuale senza decimali.

### Operatori

| Famiglia | Python | In Java |
| --- | --- | --- |
| Aritmetici | `+` `-` `*` `/` | uguali |
| Confronto | `==` `!=` `<` `<=` `>` `>=` | uguali |
| Logici | `and` `or` `not` | `&&` `\|\|` `!` |

| Operatore | Che cosa fa | Esempio | Risultato |
| --- | --- | --- | --- |
| `/` | divisione, sempre `float` | `7 / 2` | `3.5` |
| `//` | divisione intera | `7 // 2` | `3` |
| `%` | resto | `7 % 2` | `1` |
| `**` | potenza | `2 ** 10` | `1024` |

⚠️ <u>In Java `7 / 2` dà 3. In Python dà 3.5.</u> `+` e `*` funzionano anche sui testi: `"3" + "4"` dà `'34'` e `"ciao" * 3` dà `'ciaociaociao'`.

### Tre famiglie di errori

| Famiglia | Quando | Esempio |
| --- | --- | --- |
| **Sintassi** | prima di partire | `print(vita - danno` (parentesi mai chiusa) |
| **Esecuzione** | mentre gira | sommare testo e numero; dividere per zero |
| **Logica** | mai: gira, ma sbaglia | `media = forza + bonus / 2` |

L'errore di logica è il più difficile: Python non protesta, la risposta è sbagliata. Si trova **prevedendo il risultato a mano** e confrontando. Un **traceback** si legge dal basso: l'ultima riga dice *che cosa* è successo, le righe sopra dicono *dove*.

# 🎲 Parte 2 · Decisioni, cicli e funzioni (S2)

## 📘 Ripasso: la teoria che ti serve

### L'indentazione è il blocco

In Java un blocco era racchiuso fra `{ }`. In Python <u>l'indentazione *è* il blocco</u>: le righe spostate a destra dopo i due punti `:` appartengono a quel blocco. Lo standard è di **quattro spazi**.

```python
vita = 0
if vita > 0:
    print("Aria combatte.")    # dentro il blocco: solo se vita > 0
print("Fine del turno.")       # fuori dal blocco: sempre
```

### `if`, `elif`, `else`

| In Java | In Python |
| --- | --- |
| `if (...) { ... }` | `if ...:` con blocco indentato |
| `else if (...)` | `elif ...:` |
| `else { ... }` | `else:` |
| `&&`, `\|\|`, `!` | `and`, `or`, `not` |

Come in Java, vince il **primo ramo vero**. In un `if` serve il confronto `==`, non l'assegnazione `=` (che dà `SyntaxError`).

```python
dado = 4
if dado == 6:
    print("Colpo critico!")
elif dado >= 3:
    print("Colpito.")
else:
    print("Mancato.")        # stampa: Colpito.
```

### `while`, `for` e `range`

`while` controlla la condizione prima di ogni giro. Manca `++`: si scrive `turno += 1`.

```python
vita_goblin = 30
turno = 0
while vita_goblin > 0:
    turno += 1
    vita_goblin -= 8       # dopo 4 turni: -2 (la vita negativa è una scelta NON gestita dall'algoritmo)
```

Il `for` di Python prende gli elementi di una sequenza uno alla volta. `range` fornisce una sequenza di interi, **l'ultimo escluso**:

| `range(...)` | Valori | In Java |
| --- | --- | --- |
| `range(5)` | 0, 1, 2, 3, 4 | `for (int i = 0; i < 5; i++)` |
| `range(1, 7)` | 1, 2, 3, 4, 5, 6 | `for (int i = 1; i < 7; i++)` |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 | `for (int i = 0; i < 10; i += 2)` |
| `range(3, 0, -1)` | 3, 2, 1 | `for (int i = 3; i > 0; i--)` |

`break` esce subito dal ciclo; `continue` salta il resto del giro e passa al successivo. Non esiste il `do ... while`: si usa `while True:` con un `break`.

```python
for tentativo in range(1, 8):
    if tentativo % 2 == 0:
        continue                       # i pari non contano
    if tentativo == 5:
        print("La porta si apre.")
        break                          # esco dal ciclo
    print("La porta resiste al tentativo", tentativo)
```

### Funzioni: `def`, `return`, valori predefiniti

```python
def attacco(forza, tiro):
    return forza + tiro

danno = attacco(7, 4)    # 11
```

`def` crea la funzione **senza eseguirla**: il corpo parte con la chiamata. Si usa `snake_case` per i nomi (PEP 8).

<u>`print` mostra, `return` consegna.</u> Una funzione senza `return` restituisce `None`: lo schermo non è una variabile.

```python
def danno_stampato(forza):
    print(forza * 2)      # stampa e basta: restituisce None
x = danno_stampato(7)     # x vale None; x + 1 darebbe TypeError
```

Un parametro può avere un **valore predefinito**; nella chiamata si può scegliere un parametro per nome. Due `def` con lo stesso nome **non** fanno overloading: la seconda sostituisce la prima.

```python
def danno_subito(colpo, armatura=2):
    return max(0, colpo - armatura)   # max(0, ...) evita danni negativi

danno_subito(7)                  # 5
danno_subito(7, armatura=5)      # 2
```

### Dove vive una variabile

Una variabile **locale** appartiene a una chiamata; una **globale** è fuori dalle funzioni. Se dentro una funzione **assegni** a un nome, quel nome diventa locale *in tutta la funzione*:

```python
vita = 100
def cura():
    vita = vita + 20      # UnboundLocalError: 'vita' è locale e non ha ancora un valore
```

<u>Preferisci parametri e `return`</u>: i dati entrano come parametri e escono con `return`; poi è il chiamante a spostare la freccia: `vita = cura(vita)`. (`global vita` funzionerebbe, ma nasconde la modifica.)

### Moduli

Un **modulo** è un'unità di codice importabile, spesso un file `.py`. `import random` poi `random.randint(1, 6)` dà un intero da 1 a 6, **estremi compresi** (`range(1, 6)`, invece, si ferma a 5). Con `from math import sqrt` importi solo il nome che ti serve. Python esegue le righe del modulo al primo import; `if __name__ == "__main__":` separa il programma vero dalle definizioni riutilizzabili.

### 🤓 Le funzioni sono oggetti e la trappola del valore predefinito

Una funzione è un valore come gli altri: `azione = attacco` dà una seconda freccia verso la stessa funzione, e `attacco(7, 4)` (con le parentesi) la **chiama**. Si può quindi passare una funzione come **parametro** a un'altra funzione.

Un valore predefinito viene creato **una volta sola**, quando Python esegue il `def`: con una lista, tutte le chiamate condividono la stessa lista!

```python
def aggiungi(oggetto, zaino=[]):     # TRAPPOLA
    zaino.append(oggetto)
    return zaino
aggiungi("spada")      # ['spada']
aggiungi("pozione")    # ['spada', 'pozione']  <- sorpresa!
```

Il rimedio idiomatico: `zaino=None` e dentro `if zaino is None: zaino = []`.

### Liste

Una **lista** è <u>una fila di caselle numerate, che parte da zero</u>. Si scrive fra parentesi quadre. Gli indici negativi contano dalla fine.

```python
zaino = ["spada", "pozione", "torcia"]
print(zaino[0])     # 'spada'   -> primo elemento
print(zaino[-1])    # 'torcia'  -> ultimo elemento
print(len(zaino))   # 3
```

```text
indice:     0          1          2
         ┌────────┬──────────┬─────────┐
zaino ──▶│ "spada"│ "pozione"│ "torcia"│
         └────────┴──────────┴─────────┘
indice:    -3         -2         -1
```

| Operazione | Che cosa fa | In Java (`ArrayList`) |
| --- | --- | --- |
| `zaino.append(x)` | aggiunge in fondo | `zaino.add(x)` |
| `zaino.insert(i, x)` | inserisce in posizione `i` | `zaino.add(i, x)` |
| `zaino.remove(x)` | toglie il primo `x` | `zaino.remove(x)` |
| `zaino.pop()` | toglie e **restituisce** l'ultimo | `zaino.remove(zaino.size() - 1)` |
| `zaino[i]` | legge la casella `i` | `zaino.get(i)` |
| `zaino[i] = x` | sostituisce la casella `i` | `zaino.set(i, x)` |
| `x in zaino` | `True` se c'è | `zaino.contains(x)` |
| `len(zaino)` | quanti elementi | `zaino.size()` |

⚠️ Una casella che non esiste dà `IndexError: list index out of range`.

### Scorrere una lista

`for oggetto in zaino` dà gli **elementi**, non gli indici. Se servono entrambi: `for numero, oggetto in enumerate(zaino, start=1):`.

### Somme, massimi, ordinamenti

```python
danni = [8, 12, 5, 9]
sum(danni)        # 34
max(danni)        # 12
min(danni)        # 5
sorted(danni)     # [5, 8, 9, 12] -> crea una NUOVA lista; danni resta intatta
danni.sort()      # ordina la lista stessa e restituisce None
```

### Slicing

`lista[inizio:fine]` prende una parte; <u>la fine non è compresa</u>. Restituisce sempre una **nuova** lista.

```python
zaino = ["spada", "pozione", "torcia", "corda", "mappa"]
zaino[1:3]    # ['pozione', 'torcia']
zaino[:2]     # ['spada', 'pozione']
zaino[3:]     # ['corda', 'mappa']
zaino[-2:]    # ['corda', 'mappa']
zaino[::-1]   # al contrario
```

Il terzo numero è il **passo**: `zaino[::2]` prende un elemento ogni due. I **testi** sono sequenze come le liste (indici, `len`, slicing) ma sono **immutabili**: `nome[0] = "I"` dà `TypeError`.

### Copie e frecce

<u>`b = a` crea una seconda freccia verso la **stessa** lista</u> (*aliasing*). Per una copia indipendente: `a.copy()`, `a[:]` o `list(a)`.

```text
zaino_aria ──┐
             ├──▶ ["spada", "pozione", "torcia"]     (b = a: una lista sola)
zaino_kofi ──┘
```

Una **tupla** è una sequenza **immutabile**, fra parentesi tonde: `posizione = (2, 1)`. Permette anche di assegnare più variabili in una riga: `riga, colonna = 1, 1`.

### Matrici

Una **matrice** è <u>una lista che contiene liste</u>: ogni lista interna è una riga. Per leggere una casella servono due indici: <u>prima la riga, poi la colonna</u>.

```python
mappa = [
    ["#", "#", "#", "#", "#", "#"],
    ["#", "@", ".", ".", "M", "#"],
    ["#", ".", "#", ".", "$", "#"],
    ["#", "#", "#", "#", "#", "#"],
]
mappa[1][4]        # 'M'  -> riga 1, colonna 4
len(mappa)         # 4 righe
len(mappa[0])      # 6 colonne
```

`#` muro, `.` pavimento, `@` l'eroe, `M` un mostro, `$` il tesoro. Per visitare ogni casella servono **due cicli annidati**:

```python
for riga in mappa:
    for cella in riga:
        print(cella, end=" ")   # end=" " al posto dell'andata a capo
    print()                     # a capo a fine riga

for r in range(len(mappa)):                # quando servono le coordinate
    for c in range(len(mappa[r])):
        if mappa[r][c] == "$":
            print("Tesoro in riga", r, "colonna", c)
```

**Muovere l'eroe**: si calcola la casella di arrivo, si controlla che non sia un muro, poi si cambiano **due** caselle (la partenza torna `.`, l'arrivo diventa `@`). Un passo in alto è `riga - 1`, in basso `riga + 1`, a sinistra `colonna - 1`.

### 🤓 La matrice traditrice

```python
griglia = [[0] * 3] * 3             # TRAPPOLA: tre frecce verso la STESSA riga

griglia_ok = [[0] * 3 for r in range(3)]   # forma sicura: una riga nuova a ogni giro
```

# 📇 Parte 4 · Dizionari (S4)

## 📘 Ripasso: la teoria che ti serve

### Il Google di carta

Nel 1895, a Bruxelles, Paul Otlet e Henri La Fontaine costruirono il **Mundaneum**: milioni di schede in cassetti di legno, ognuna con un'intestazione in alto e le informazioni sotto. <u>Non cercavi «la scheda numero 4.211.087»: cercavi la scheda che si chiama in un certo modo.</u> È l'idea del dizionario.

### Chiave e valore

Un **dizionario** è una raccolta di coppie **chiave → valore**. <u>Si cerca un dato dal suo nome (la chiave), non dalla sua posizione.</u> Si scrive fra graffe, con i due punti.

```python
scheda = {"nome": "Aria", "vita": 100, "forza": 7}
print(scheda["vita"])   # 100
```

| | Lista | Dizionario |
| --- | --- | --- |
| Si scrive | `["Aria", 100, 7]` | `{"nome": "Aria", "vita": 100, "forza": 7}` |
| Si legge con | un indice: `scheda[1]` | una chiave: `scheda["vita"]` |
| Esempio | lo zaino, i turni, la mappa | la scheda, la rubrica, l'inventario |

### Modificare, aggiungere, togliere

```python
scheda = {"nome": "Aria", "vita": 100, "forza": 7}
scheda["vita"] = 85         # chiave già presente: modifica
scheda["livello"] = 2       # chiave nuova: aggiunge
len(scheda)                 # 4 coppie
del scheda["livello"]       # rimuove la coppia
```

<u>La stessa scrittura `scheda[chiave] = valore` modifica una chiave che c'è e aggiunge una chiave che non c'è.</u> Le chiavi sono quasi sempre testi o numeri; i valori possono essere qualunque cosa, anche liste e altri dizionari.

### Chiavi che non ci sono

⚠️ <u>Una chiave che non esiste non dà «niente»: dà un `KeyError`.</u> Due rimedi: chiedere prima con `in`, oppure usare `get`.

```python
"oro" in scheda          # False
scheda.get("oro")        # None -> non crasha
scheda.get("oro", 0)     # 0    -> valore di riserva
scheda[0]                # KeyError: 0 -> i dizionari NON hanno indici numerici automatici
```

### Scorrere un dizionario

Un `for` su un dizionario dà le **chiavi**. Per chiave e valore insieme: `.items()`.

```python
for chiave in scheda:
    print(chiave)
for chiave, valore in scheda.items():
    print(f"{chiave}: {valore}")
```

| Metodo | Che cosa dà |
| --- | --- |
| `scheda.keys()` | le chiavi |
| `scheda.values()` | i valori |
| `scheda.items()` | le coppie (chiave, valore), come tuple |

### L'inventario: contare con un dizionario

```python
inventario = {"pozione": 3, "freccia": 20}
inventario["pozione"] -= 1                                  # usa una pozione
inventario["chiave"] = inventario.get("chiave", 0) + 1      # funziona anche se "chiave" non c'era
sum(inventario.values())                                    # 23 -> somma totale degli oggetti
```

La riga con `get` è un'espressione da ricordare: <u>«prendi quante ne ho, oppure 0 se non ne ho, e aggiungi 1»</u>.

### Una lista di dizionari è una tabella

<u>Una lista di dizionari è una tabella: ogni dizionario è una riga (un record), ogni chiave è una colonna.</u>

```python
bestiario = [
    {"nome": "Goblin", "vita": 30, "attacco": 4},
    {"nome": "Scheletro", "vita": 45, "attacco": 6},
    {"nome": "Drago", "vita": 300, "attacco": 25},
]
bestiario[2]["nome"]     # 'Drago' -> il terzo mostro, poi il suo nome
for mostro in bestiario:
    print(f"{mostro['nome']}: vita {mostro['vita']}")   # nelle f-string, apici singoli dentro
```

Schema «cercare il più pericoloso»:

```python
piu_forte = bestiario[0]
for mostro in bestiario:
    if mostro["attacco"] > piu_forte["attacco"]:
        piu_forte = mostro
```

### Dizionari che contengono liste e tuple

```python
eroe = {"nome": "Aria", "vita": 100, "zaino": ["spada", "torcia"], "posizione": (1, 1)}
eroe["zaino"][0]                  # 'spada'
eroe["zaino"].append("pozione")   # modifica la lista dentro il dizionario
eroe["posizione"][0]              # 1
```

### Anche qui, frecce

<u>Un dizionario contiene frecce, come una lista.</u> Se prendi un mostro dal bestiario e lo modifichi, **modifichi il bestiario**:

```python
goblin = bestiario[0]
goblin["vita"] -= 10      # bestiario[0]["vita"] è cambiata!
```

Per una scheda indipendente: `bestiario[0].copy()`.

### Lista o dizionario?

| Se ti chiedi... | Usa | Esempio |
| --- | --- | --- |
| «qual è il terzo?», «in che ordine?» | lista | turni, zaino, classifica |
| «quanto vale *questa cosa*?» | dizionario | scheda, inventario, impostazioni |
| «tante cose con gli stessi campi» | lista di dizionari | bestiario, registro di classe |
| «una griglia» | lista di liste | mappa, scacchiera |

### 🤓 Chiavi immutabili, insiemi e JSON

Le chiavi devono essere **immutabili** (testi, numeri, tuple; una lista no: `TypeError: unhashable type`).
