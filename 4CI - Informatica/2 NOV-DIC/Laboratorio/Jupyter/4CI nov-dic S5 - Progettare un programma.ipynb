1. **In alto**: `import` e **costanti** (valori che non cambiano, in MAIUSCOLO: `VITA_MASSIMA = 100`);

### ✔️ `assert`, test e casi limite
«Non dà errori» non vuol dire «funziona»: il goblin a −2 punti vita era un errore di logica. `assert` controlla che una condizione sia vera, altrimenti ferma il programma con `AssertionError`. Un insieme di `assert` è un **test**: un piccolo programma che controlla un altro programma. Si rieseguono dopo ogni modifica, per controllare di non aver rotto qualcosa che funzionava.

Un **caso limite** è un valore ai bordi: zero, il massimo, una lista vuota, un danno più grande della vita. <u>Gli errori si nascondono quasi sempre ai bordi.</u>

| Funzione | Caso normale | Casi limite da provare |
| --- | --- | --- |
| `subisci_danno(vita, danno)` | `(100, 30)` | danno 0; danno uguale alla vita; danno più grande della vita |
| `cura(vita, punti)` | `(50, 30)` | cura che supera il massimo; vita già al massimo |
| una funzione che riceve il bestiario | tre mostri | bestiario vuoto; un solo mostro |

Fuori da Jupyter i test stanno in un file `test_qualcosa.py` che importa le funzioni dal tuo modulo (`from dungeon import subisci_danno, cura`) e si esegue con `python -m pytest -q`: pytest esegue ogni funzione `test_...` e riassume i risultati.

